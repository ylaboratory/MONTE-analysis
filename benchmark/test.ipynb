```python
import pandas as pd
```

```python
meta = pd.read_parquet('/grain/mk98/cancer-methyl/TCGA_Methylation_450K/all_probes_no_impute/merged_meta_with_purity.parquet')
```

```python
meta
```

```python
import seaborn as sns

import matplotlib.pyplot as plt

df = meta.dropna(subset=['Cancer.type', 'CPE'])
order = df.groupby('Cancer.type')['CPE'].median().sort_values().index

plt.figure(figsize=(14,6))
sns.boxplot(x='Cancer.type', y='CPE', data=df, order=order)
plt.xticks(rotation=90)
plt.xlabel('Cancer type')
plt.ylabel('CPE')
plt.title('CPE by Cancer type (boxplot)')
plt.tight_layout()
```
